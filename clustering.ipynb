# Segment the Data: K-Means vs DBSCAN
Run the cells top to bottom. Upload your CSV when prompted.

## 1. Load and scale

```
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN
from sklearn.decomposition import PCA

up = files.upload()
df = pd.read_csv(next(iter(up)))
num = df.select_dtypes('number').dropna()
X = StandardScaler().fit_transform(num)
num.shape
```

## 2. Elbow method

```
ks = range(1, 11)
inertia = [KMeans(k, n_init=10, random_state=42).fit(X).inertia_ for k in ks]

plt.plot(ks, inertia, 'o-')
plt.xlabel('k')
plt.ylabel('Inertia')
plt.title('Elbow Method')
plt.show()
```

## 3. K-Means
Set `k` to the elbow point from the plot above.

```
k = 3
km = KMeans(k, n_init=10, random_state=42).fit_predict(X)
pd.Series(km).value_counts().sort_index()
```

## 4. DBSCAN
Label -1 means noise. If you get one giant cluster or mostly noise, adjust `eps` (try 1 to 3).

```
db = DBSCAN(eps=2, min_samples=5).fit_predict(X)
pd.Series(db).value_counts().sort_index()
```

## 5. Compare the groupings

```
pd.crosstab(km, db, rownames=['KMeans'], colnames=['DBSCAN'])
```

## 6. PCA to 2D and scatter plots

```
pcs = PCA(n_components=2).fit_transform(X)

fig, ax = plt.subplots(1, 2, figsize=(12, 5))
for a, labels, name in zip(ax, [km, db], ['K-Means', 'DBSCAN']):
    s = a.scatter(pcs[:, 0], pcs[:, 1], c=labels, cmap='viridis', s=40)
    a.set_title(name)
    a.set_xlabel('PC1')
    a.set_ylabel('PC2')
    a.legend(*s.legend_elements(), title='Cluster')
plt.show()
```

## 7. Cluster profiles

```
print(num.groupby(km).mean().round(2))
print(num.groupby(db).mean().round(2))
```
